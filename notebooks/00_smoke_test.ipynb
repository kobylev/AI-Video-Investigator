## 8. Ping Gemini API (Optional)

**Note:** This cell will only run if `GEMINI_API_KEY` is set. It sends a minimal API request to verify connectivity.

```python
api_key = os.getenv("GEMINI_API_KEY")

if api_key:
    try:
        import google.generativeai as genai
        
        genai.configure(api_key=api_key)
        model = genai.GenerativeModel('gemini-1.5-pro-latest')
        
        # Send a minimal test prompt
        response = model.generate_content("Say 'ready' if you are online.")
        
        print("✅ Gemini API connection successful")
        print(f"   Response: {response.text.strip()}")
    except Exception as e:
        print("❌ Gemini API call failed:", e)
else:
    print("⏭️  Skipping Gemini API test (GEMINI_API_KEY not set)")
    print("   To enable: export GEMINI_API_KEY=your_api_key")
```
